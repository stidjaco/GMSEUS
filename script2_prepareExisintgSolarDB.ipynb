# Prepare Array Polygon Datasets

## GM-SEUS: Version Specific Updates

### Newly Digitized and Georeferenced Arrays (v2.0)

```python
 # Load all gdf (geojson) files in the GM-SEUS digGeorefArrays_StatePath folder and concatenate into a single geodataframe
gmseusDigGeoref = gu.load_all_gdf(gmseusDigGeorefArrays_StatePath, 'geojson', toCRS)

# Subset dataset to aoi
gmseusDigGeoref = gmseusDigGeoref[gmseusDigGeoref.intersects(unary_union(aoi.geometry))]

# Ensure no self-overlap (removal of duplicate arrays)
gmseusDigGeoref = gu.dropSelfOverlapGDF(gmseusDigGeoref)

# Check for errant geometries (empty, invalid, non-polygon)
gmseusDigGeoref = gu.checkArrayGeometries(gmseusDigGeoref)

# Drop initial area column and calculate area in m2
gmseusDigGeoref = gmseusDigGeoref.drop(columns=['area'], errors='ignore')
gmseusDigGeoref['area'] = gmseusDigGeoref['geometry'].area.round(2)

# Assume modType is c-si, and set empty columns for all required fields not provided in dataset
gmseusDigGeoref['modType'] = 'c-si'
gmseusDigGeoref['capMWAC'] = np.nan
gmseusDigGeoref['grndCvr'] = np.nan
gmseusDigGeoref['tilt'] = np.nan

# Set Source
gmseusDigGeoref['Source'] = 'GMSEUSgeoref_v2_0'

# Format data
gmseusDigGeoref = gu.formatDf(df = gmseusDigGeoref, nativeIdentifier = 'nativeID', installationYear = 'instYr', capacityMWdc = 'cap_mw', capacityMWac = 'capMWAC', area_m2 = 'area', moduleType = 'modType', groundCover = 'grndCvr', azimuth = 'azimuth', tilt = 'tilt', mountTechnology = 'mount', source = 'Source')

# Print the number of unique arrays
print(f'Number of arrays in newly georeferenced GMSEUS in the {countryName} is {len(gmseusDigGeoref)}')

# Print the total area in km2 of the arrays
print(f'Total area of arrays in newly georeferenced GMSEUS in the {countryName} is {(gmseusDigGeoref["area"].sum() / 1e6).round(2)} km2')

# Export to shapefile
gmseusDigGeoref.to_file(os.path.join(derivedTemp_path, r'gmseusDigGeoref_poly.shp'))
```

### Arrays from Newly Digitized Panel-Rows (v2.0)

```python
# Call GM-SEUS newly digitized panel-rows
gmseusDigArraysPanels = gpd.read_file(gmseusDigPanels_path)

# Transform to toCRS
gmseusDigArraysPanels = gmseusDigArraysPanels.to_crs(toCRS)

# Create arrays from panel rows
gmseusDigArraysPanels = gu.createArrayFromPanels(gmseusDigArraysPanels, panelArrayBuff, '', '', False)

# Ensure no self-overlap (removal of duplicate arrays)
gmseusDigArraysPanels = gu.dropSelfOverlapGDF(gmseusDigArraysPanels)

# Set nativeID to unique index
gmseusDigArraysPanels['nativeID'] = range(1, len(gmseusDigArraysPanels) + 1)

# Get the mount type, azimuth, length ratio, area ratio, short edge, and long edge for each panel
gmseusDigArraysPanels[['mount', 'azimuth', 'lengthRatio', 'shortEdge', 'longEdge']] = gmseusDigArraysPanels.apply(gu.assignMountType, axis=1, result_type='expand')
gmseusDigArraysPanels = gmseusDigArraysPanels.drop(columns=['lengthRatio', 'shortEdge', 'longEdge'])

# Drop initial area column and calculate area in m2
gmseusDigArraysPanels = gmseusDigArraysPanels.drop(columns=['area'], errors='ignore')
gmseusDigArraysPanels['area'] = gmseusDigArraysPanels['geometry'].area.round(2)

# Assume modType is c-si, and set empty columns for all required fields not provided in dataset
gmseusDigArraysPanels['modType'] = 'c-si'
gmseusDigArraysPanels['instYr'] = np.nan
gmseusDigArraysPanels['capMWDC'] = np.nan
gmseusDigArraysPanels['capMWAC'] = np.nan
gmseusDigArraysPanels['grndCvr'] = np.nan
gmseusDigArraysPanels['tilt'] = np.nan

# Set Source
gmseusDigArraysPanels['Source'] = 'GMSEUSdigArraysPanels_v2_0'

# Format data
gmseusDigArraysPanels = gu.formatDf(df = gmseusDigArraysPanels, nativeIdentifier = 'nativeID', installationYear = 'instYr', capacityMWdc = 'capMWDC', capacityMWac = 'capMWAC', area_m2 = 'area', moduleType = 'modType', groundCover = 'grndCvr', azimuth = 'azimuth', tilt = 'tilt', mountTechnology = 'mount', source = 'Source')

# Print the number of unique arrays
print(f'Number of arrays created by digitized panel-rows in {countryName} is {len(gmseusDigArraysPanels)}')

# Print the total area in km2 of the arrays
print(f'Total area of arrays created by digitized panel-rows in {countryName} is {(gmseusDigArraysPanels["area"].sum() / 1e6).round(2)} km2')

# Export to shapefile
gmseusDigArraysPanels.to_file(os.path.join(derivedTemp_path, r'gmseusDigArraysPanels_poly.shp'))
```

### Generate GM-SEUS comission datasets: Known rooftop dataset and known non-solar array objects

```python
# Call GM-SEUS prior version (used to digitize and georeference arrays)
gmseusPriorVersion = gpd.read_file(gmseusPriorVersion_path)
gmseusPriorVersion_pointsToDigitize = gpd.read_file(gmseusPriorVersionPointsToDigitize_path)

# Transform both dataframes to desired CRS
gmseusPriorVersion = gmseusPriorVersion.to_crs(toCRS)
gmseusPriorVersion_pointsToDigitize = gmseusPriorVersion_pointsToDigitize.to_crs(toCRS)

# Buffer prior version and points to digitize to account for slight mismatches in geometry
gmseusPriorVersion_PTDbuff = gu.bufferFacilityPointData(gmseusPriorVersion_pointsToDigitize, overlapDist)

# Call metadata dataframes from digitization and georeferencing notes (csv)
gmseusDigPanels_notes = pd.read_csv(gmseusDigPanels_notesPath)
gmseusDigGeorefArrays_notes = pd.read_csv(gmseusDigGeorefArrays_notesPath)

# From each notes page, get Source, nativeID, and Rooftop columns
gmseusDigPanels_notes = gmseusDigPanels_notes[['Source', 'nativeID', 'Rooftop']]
gmseusDigGeorefArrays_notes = gmseusDigGeorefArrays_notes[['Source', 'nativeID', 'Rooftop']]

# Select only rows where Rooftop is 'yes'
gmseusDigPanels_notesRooftop = gmseusDigPanels_notes[gmseusDigPanels_notes['Rooftop'].str.lower() == 'yes']
gmseusDigGeorefArrays_notesRooftop = gmseusDigGeorefArrays_notes[gmseusDigGeorefArrays_notes['Rooftop'].str.lower() == 'yes']

# Get list of Source and nativeID combinations for rooftop arrays from both notes dataframes
rooftopIDs = pd.concat([gmseusDigPanels_notesRooftop[['Source', 'nativeID']], gmseusDigGeorefArrays_notesRooftop[['Source', 'nativeID']]], ignore_index=True)

# Print the number of unique rooftop arrays from notes
print(f'Number of rooftop arrays identified in digitization and georeferencing notes in the {countryName} is {len(rooftopIDs)}')

# Pull rooftop arrays from prior GM-SEUS version (DigPanels) and from prior points to digitize (DigGeoref) based on Source and nativeID combinations
gmseusPriorVersion_DigRooftop = gmseusPriorVersion.merge(gmseusDigPanels_notesRooftop, on=['Source', 'nativeID'], how='inner')
gmseusPriorVersion_GeorefRooftop = gmseusPriorVersion_PTDbuff.merge(gmseusDigGeorefArrays_notesRooftop, on=['Source', 'nativeID'], how='inner')

# Combine both rooftop dataframes as geodataframes
gmseusPriorVersion_rooftop = pd.concat([gmseusPriorVersion_DigRooftop, gmseusPriorVersion_GeorefRooftop], ignore_index=True)
gmseusPriorVersion_rooftop = gpd.GeoDataFrame(gmseusPriorVersion_rooftop, geometry='geometry', crs=toCRS)

# Schema changes slightly from preprocessing to final formatting for GMSEUS v1.0. 
# capMW and cap_mw are the same, totArea and area are the same, avgAzimuth and azimuth are the same, and arrayID and nativeID are the same (at least for this purpose)
gmseusPriorVersion_rooftop['capMW'] = gmseusPriorVersion_rooftop['capMW'].fillna(gmseusPriorVersion_rooftop['cap_mw'])
gmseusPriorVersion_rooftop['totArea'] = gmseusPriorVersion_rooftop['area'].fillna(gmseusPriorVersion_rooftop['area'])
gmseusPriorVersion_rooftop['avgAzimuth'] = gmseusPriorVersion_rooftop['avgAzimuth'].fillna(gmseusPriorVersion_rooftop['azimuth'])
gmseusPriorVersion_rooftop['arrayID'] = gmseusPriorVersion_rooftop['arrayID'].fillna(gmseusPriorVersion_rooftop['nativeID'])

# Drop columns no longer needed
gmseusPriorVersion_rooftop = gmseusPriorVersion_rooftop.drop(columns=['cap_mw', 'area', 'azimuth', 'nativeID'], errors='ignore')

# Set empty columns for all required fields not provided in dataset
gmseusPriorVersion_rooftop['capMWAC'] = np.nan
gmseusPriorVersion_rooftop['grndCvr'] = np.nan
gmseusPriorVersion_rooftop['tilt'] = np.nan

# Format data
gmseusPriorVersion_rooftop = gu.formatDf(df = gmseusPriorVersion_rooftop, nativeIdentifier = 'arrayID', installationYear = 'instYr', capacityMWdc = 'capMW', capacityMWac = 'capMWAC', area_m2 = 'totArea', moduleType = 'modType', groundCover = 'grndCvr', azimuth = 'avgAzimuth', tilt = 'tilt', mountTechnology = 'mount', source = 'Source')

# Print the number of unique rooftop arrays
print(f'Number of rooftop arrays in prior GMSEUS version in the {countryName} is {len(gmseusPriorVersion_rooftop)}')

# Export to shapefile
gmseusPriorVersion_rooftop.to_file(os.path.join(derivedTemp_path, r'gmseusPriorVersionRooftop_poly.shp'))
```

## United States Photovoltaic Database (USPVDB)

```python
# Call data
uspvdb = gpd.read_file(uspvdb_path)

# Transform to desired projection
uspvdb = uspvdb.to_crs(toCRS)

# Subset dataset to aoi
uspvdb = uspvdb[uspvdb.intersects(unary_union(aoi.geometry))]

# Ensure no self-overlap (removal of duplicate arrays)
uspvdb = gu.dropSelfOverlapGDF(uspvdb)

# Set a mount column. 
# If p_axis is 'single-axis', set to single_axis. If p_axis is 'fixed-axis', set to fixed_axis. If p_axis is 'dual-axis', set to dual_axis. If p_axis is anything else, set to mixed (not not use notin)
uspvdb['mount'] = np.nan
uspvdb['mount'] = uspvdb['mount'].astype(object)
uspvdb.loc[uspvdb['p_axis'] == 'single-axis', 'mount'] = 'single_axis'
uspvdb.loc[uspvdb['p_axis'] == 'fixed-tilt', 'mount'] = 'fixed_axis'
uspvdb.loc[uspvdb['p_axis'] == 'dual-axis', 'mount'] = 'dual_axis'
uspvdb.loc[~uspvdb['p_axis'].isin(['single-axis', 'fixed-tilt', 'dual-axis']), 'mount'] = 'mixed'

# Set empty columns for all required fields not provided in dataset
uspvdb['grndCvr'] = np.nan

# Set Source
uspvdb['Source'] = 'USPVDB'

# Save a dataframe copy of p_sys_type = rooftop, then drop from main dataframe
uspvdb_rooftop = uspvdb[uspvdb['p_sys_type'] == 'rooftop']
uspvdb = uspvdb[uspvdb['p_sys_type'] != 'rooftop']

# Format data
uspvdb = gu.formatDf(df = uspvdb, nativeIdentifier = 'case_id', installationYear = 'p_year', capacityMWdc = 'p_cap_dc', capacityMWac = 'p_cap_ac', area_m2 = 'p_area', moduleType = 'p_tech_sec', groundCover = 'grndCvr', azimuth = 'p_azimuth', tilt = 'p_tilt', mountTechnology = 'mount', source = 'Source')
uspvdb_rooftop = gu.formatDf(df = uspvdb_rooftop, nativeIdentifier = 'case_id', installationYear = 'p_year', capacityMWdc = 'p_cap_dc', capacityMWac = 'p_cap_ac', area_m2 = 'p_area', moduleType = 'p_tech_sec', groundCover = 'grndCvr', azimuth = 'p_azimuth', tilt = 'p_tilt', mountTechnology = 'mount', source = 'Source')

# ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~ Export

# Print the number of rooftop arrays
print(f'Number of rooftop arrays in USPVDB in the {countryName} is {len(uspvdb_rooftop)}')

# Print the total area in km2 of the rooftop arrays
print(f'Total area of rooftop arrays in USPVDB in the {countryName} is {(uspvdb_rooftop['geometry'].area.round(2).sum() / 1e6).round(2)} km2')

# Export to shapefile
uspvdb_rooftop.to_file(os.path.join(derivedTemp_path, r'uspvdbRooftop_poly.shp'))

# Print the number of unique arrays
print(f'Number of arrays in USPVDB in the {countryName} is {len(uspvdb)}')

# Print the total area in km2 of the arrays
print(f'Total area of arrays in USPVDB in the {countryName} is {(uspvdb["area"].sum() / 1e6).round(2)} km2')

# Export to shapefile
uspvdb.to_file(os.path.join(derivedTemp_path, r'uspvdb_poly.shp'))
```

# Create an Existing Solar Array Dataset with Preferential Selection of Spatial and Metadata Information
Here, we check existing solar databases for high-quality metadata. The order of preference for each attribute is distinct based on the percieved quality of the derivation or acquistion method of that attribute. 
* Dataset attribute contributions are ordered by percieved quality, and thus attribution ranking to the final GM-SEUS dataset.
* Arrays from manual georeferencing and digitization (Source: 'GMSEUSgeoref_v2_0'), take the lowest order of preference from all datasets, but fills necessary gaps.
* We do not include attributes from all datasets for quality control purposes, even if there are no high-quality dataset attribution. We do this because unless the metadata source was derived from perimt data or high-quality sub-array metadata extraction, we assume our methods used here are more accurate. 
    * Some datasets have extensive methods for certain attributes (e.g., GEM GSPT derivation of capacity is array by array intensive investiation of web sources), despite not having all permit level data. 
    * We denote dataset attributes that exist but that we do not use by <span style="background-color: #FF2800">*color*</span> and explain in line.
* Metadata where no quality source yet exists for attribution (ground cover ratio, module efficiency) are omitted here and assess later in the codebase

**Installation Year** 
* USPVDB: `oper_year`
* InSPIRE: `Year Installed`
* LBNL-USS: `Solar COD Year`
* SolarPACES: `Year_operational`
* GSPT: `Start year`
* GPPDB: `commissioning_year`
* OSM: `start_date`
* SAM: `constructed_before` & `constructed_after` -- derived from Sentinel-2, average taken, dropped any year prior to 2018 to prevent first year of S2 bias
* CWSD: `year_right` -- derived from Sentinel-2, dropped any year prior to 2018 to prevent first year of S2 bias
* <span style="background-color: #FF2800">*CCVPV*</span>: `inst_Yr` -- We improve our our prior LandTrendr methods, so omit

**Installed Capacity (DC)**
* USPVDB: `p_cap_dc`
* InSPIRE: `System Size`
* LBNL-USS: `Solar Capacity MW-DC`
* PV-DAQ: `system_size`
* GSPT: `Capacity (MW)` -- DC and AC mixed (differentiaed in above code)
* GPPDB: `capacity_mw` -- DC? (uncertain)
* <span style="background-color: #FF2800">*CCVPV*</span>: `TPVPp` -- Derived by lower-fidelity installation year analysis, this product will be more robust
* <span style="background-color: #FF2800">*OSM*</span>: `plant:output:electricity` -- (defined as: *peak rated output capacity*) -- crowd sourced information with unclear accuracy of reporting (concerns between AC/DC, original vs expanded capacity, etc.)

**Installed Capacity (AC)**
* USPVDB: `p_cap_ac`
* LBNL-USS: `Solar Capacity MW-AC`
* SolarPaces: `Capacity_MW` -- e or electric, synonomus to AC?
* GSPT: `Capacity (MW)` -- DC and AC mixed (differentiaed in above code)
* <span style="background-color: #FF2800">*TZ-SAM*</span>: `capacity_mw` -- AC -- Derived by course Regional GCR and inverter loading ratio (ILR) and for often erronouesly large array areas (due to coarse imagery)

**Module Type** 
* SolarPaces: inherent in the dataset -- CSP -- This takes precedent for csp arrays
* USPVDB: `p_tech_sec`
* LBNL-USS: `Solar Tech Sub`
* GSPT: `Technology Type` -- only PV or Thermal delineation

**Mount Technology**
* USPVDB: `p_axis`
* InSPIRE: `Type Of Array`
* LBNL-USS: `Tracking Type`
* CCVPV: `Class`
* SolarPaces: assumed from `Technology`

**Azimuth**
* USPVDB: `p_azimuth`
* LBNL-USS: `Azimuth`
* <span style="background-color: #FF2800">*OSM*</span>: `generator:orientation` (we have not yet acquired this from OSM, need to update `script0`) -- Have not pulled this information from OSM yet, but there remains concern over quality of crowd sourced information when attribute is not clearly explained

**Module Tilt**
* USPVDB: `p_tilt`
* LBNL-USS: `Tilt`
* <span style="background-color: #FF2800">*CCVPV*</span>: `Bopt` -- 'optimum tilt' form a simplified method using latitude only, drop

**Agrivoltaic Type**
* InSPIRE: `Habitat Type` -- inherant in dataset, vegetation
* CECSFC: `Type` -- "Parking" assumed to be impervious
* <span style="background-color: #FF2800">*USPVDB*</span>: `AVtype` -- Derived from InSPIRE

```python
# Call polygon data
gmseusDigGeoref = gpd.read_file(os.path.join(derivedTemp_path, r'gmseusDigGeoref_poly.shp'))
gmseusDigArraysPanels = gpd.read_file(os.path.join(derivedTemp_path, r'gmseusDigArraysPanels_poly.shp'))
uspvdb = gpd.read_file(os.path.join(derivedTemp_path, r'uspvdb_poly.shp'))
cecsfc = gpd.read_file(os.path.join(derivedTemp_path, r'cecsfc_poly.shp'))
osm = gpd.read_file(os.path.join(derivedTemp_path, r'osm_poly.shp'))
ccvpv = gpd.read_file(os.path.join(derivedTemp_path, r'ccvpv_poly.shp'))
tzsam = gpd.read_file(os.path.join(derivedTemp_path, r'tzsam_poly.shp'))
grw = gpd.read_file(os.path.join(derivedTemp_path, r'grw_poly.shp'))
cwsd = gpd.read_file(os.path.join(derivedTemp_path, r'cwsd_poly.shp'))

# Call point data
inspirePoint = gpd.read_file(os.path.join(derivedTemp_path, r'inspire_point.shp'))
lbnlUssPoint = gpd.read_file(os.path.join(derivedTemp_path, r'lbnlUss_point.shp'))
pvdaqPoint = gpd.read_file(os.path.join(derivedTemp_path, r'pvdaq_point.shp'))
solarPacesPoint = gpd.read_file(os.path.join(derivedTemp_path, r'solarPaces_point.shp'))
gsptPoint = gpd.read_file(os.path.join(derivedTemp_path, r'gspt_point.shp'))
gppdbPoint = gpd.read_file(os.path.join(derivedTemp_path, r'gppdb_point.shp'))

# Call rooftop data (that we currently have access to, we will also filter by building footprint overlap in script 3)
gmseusRooftop = gpd.read_file(os.path.join(derivedTemp_path, r'gmseusPriorVersionRooftop_poly.shp'))
uspvdbRooftop = gpd.read_file(os.path.join(derivedTemp_path, r'uspvdbRooftop_poly.shp'))
cecsfcRooftop = gpd.read_file(os.path.join(derivedTemp_path, r'cecsfcRooftop_poly.shp'))

# Call GM-SEUS version georeferencing point connectors. Transform DigGeoref connectors to shared CRS
gmseusDigGeoref_connectors = gpd.read_file(gmseusDigGeorefArrays_PointConnectorPath)
gmseusDigGeoref_connectors = gmseusDigGeoref_connectors.to_crs(toCRS)

# Drop low confidence attributes from relevant datasets
ccvpv['instYr'] = -9999
ccvpv['capMWDC'] = -9999
ccvpv['tilt'] = -9999
osm['capMWDC'] = -9999
tzsam['capMWAC'] = -9999
uspvdb['grndCvr'] = uspvdb['grndCvr'].fillna('').astype(str)

# ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~ Create a GM-SEUS rooftop dataset for filtering (appended in script 3)

# Merge the rooftop datasets
rooftopList = [gmseusRooftop, uspvdbRooftop, cecsfcRooftop]
rooftopMerged = gu.preferentialSpatialFilter(rooftopList)

# ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~ Merge polygon datasets into a single high-spatial quality dataset

# Remove arrays with overlap to keep arrays based on order of preference (spatial quality, completeness, updatedness)
# Level of priority: gmseusDigGeoref, gmseusDigArraysPanels, uspvdb, cecsfc, osm, ccvpv, tzsam, grw, cwsd
gdfList = [gmseusDigGeoref, gmseusDigArraysPanels, uspvdb, cecsfc, osm, ccvpv, tzsam, grw, cwsd]
merged = gu.preferentialSpatialFilter(gdfList)

# Mask the merged dataset with the US boundary (this is a redundant step, is a good check)
merged = merged[merged.intersects(unary_union(aoi.geometry))]

# Drop merged arrays that overlap with rooftop arrays in rooftopMerged
merged = merged[~merged.intersects(unary_union(rooftopMerged.geometry))]

# For a collection of reasons, some geometries contain invalid geometries. We will use the gmseusUtils function to check and fix these geometries.
merged = gu.checkArrayGeometries(merged)

# Drop all columns that are not needed for the analysis
merged = merged[['nativeID', 'instYr', 'capMWDC', 'capMWAC', 'area', 'modType', 'grndCvr', 'azimuth', 'tilt', 'mount', 'Source', 'geometry']]

# ~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~~ Prepare point datasets (buffer) for intersection with merged polygon dataset

# Buffer point data by overlapDist (190 meters)
inspire = gu.bufferFacilityPointData(inspirePoint, overlapDist)
lbnlUss = gu.bufferFacilityPointData(lbnlUssPoint, overlapDist)
pvdaq = gu.bufferFacilityPointData(pvdaqPoint, overlapDist)
solarPaces = gu.bufferFacilityPointData(solarPacesPoint, overlapDist)
gspt = gu.bufferFacilityPointData(gsptPoint, overlapDist)
gppdb = gu.bufferFacilityPointData(gppdbPoint, overlapDist)
```

## Collect Highest-Confidence Metadata from all Array Datasets

```python
# Set a copy of the merged polygon dataset to mergedMetadata for metadata filling
mergedMetadata = merged.copy()

# Installation Year
instYr_metadataPriorityList = [uspvdb, inspire, lbnlUss, pvdaq, solarPaces, gspt, gppdb, osm, tzsam, grw, cwsd, gmseusDigGeoref]
mergedMetadata = gu.iterateFillMetadataByOverlap(mergedMetadata, instYr_metadataPriorityList, aggConfig = {'instYr': ('mode', 'num', True)}) # Attribute: (aggFunc, dataType, overwriteExisting)

# Capacity MWDC
capMWDC_metadataPriorityList = [uspvdb, inspire, lbnlUss, pvdaq, gspt, gppdb, gmseusDigGeoref]
mergedMetadata = gu.iterateFillMetadataByOverlap(mergedMetadata, capMWDC_metadataPriorityList, aggConfig = {'capMWDC': ('median', 'num', True)})

# Capacity MWAC
capMWAC_metadataPriorityList = [uspvdb, lbnlUss, solarPaces, gspt, gmseusDigGeoref]
mergedMetadata = gu.iterateFillMetadataByOverlap(mergedMetadata, capMWAC_metadataPriorityList, aggConfig = {'capMWAC': ('median', 'num', True)})

# Module Type
modType_metadataPriorityList = [solarPaces, uspvdb, lbnlUss, gspt, gmseusDigGeoref]
mergedMetadata = gu.iterateFillMetadataByOverlap(mergedMetadata, modType_metadataPriorityList, aggConfig = {'modType': ('mode', 'str', True)})

# Mount Technology
mount_metadataPriorityList = [uspvdb, inspire, lbnlUss, ccvpv, solarPaces, gmseusDigGeoref]
mergedMetadata = gu.iterateFillMetadataByOverlap(mergedMetadata, mount_metadataPriorityList, aggConfig = {'mount': ('mode', 'str', True)})

# Azimuth
azimuth_metadataPriorityList = [uspvdb, lbnlUss, gmseusDigGeoref]
mergedMetadata = gu.iterateFillMetadataByOverlap(mergedMetadata, azimuth_metadataPriorityList, aggConfig = {'azimuth': ('median', 'num', True)})

# Tilt
tilt_metadataPriorityList = [uspvdb, lbnlUss, gmseusDigGeoref]
mergedMetadata = gu.iterateFillMetadataByOverlap(mergedMetadata, tilt_metadataPriorityList, aggConfig = {'tilt': ('median', 'num', True)})

# Agrivoltaic Type
grndCvr_metadataPriorityList = [inspire, cecsfc, gmseusDigGeoref]
mergedMetadata = gu.iterateFillMetadataByOverlap(mergedMetadata, grndCvr_metadataPriorityList, aggConfig = {'grndCvr': ('mode', 'str', True)})
```

## Final Checks and Export Metadata-Rich Existing Array Dataset

```python
# ~~~~~~~~~~~~~~~~~~~~~~~~~ Rooftop Array Data

# Print the number of rooftop arrays
print(f'The number of rooftop arrays in the merged rooftop dataset is {len(rooftopMerged)}')

# Print the total area in km2 of the rooftop arrays
print(f'The total area of rooftop arrays in the merged rooftop dataset is {(rooftopMerged["area"].sum() / 1e6).round(2)} km2')

# Export shapefile
rooftopMerged.to_file(os.path.join(derivedTemp_path, r'existingRooftopArrayShapes.shp'))

# ~~~~~~~~~~~~~~~~~~~~~~~~~ Point Array Data

# Get point data that is not within 190 meters of other point data. Spatial quality is in theory the same for these, given percieved dataset quality
# Order of priority: inspire, lbnlUss, pvdaq, gspt, solarPaces, gppdb
gdfList_points = [inspire, lbnlUss, pvdaq, gspt, solarPaces, gppdb]
mergedPoints = gu.preferentialSpatialFilter(gdfList_points)

# Get points that are not within 190 meters of merged arrays
pointsUnique = mergedPoints[~mergedPoints.intersects(unary_union(mergedMetadata.geometry))]

# Get points that are not rooftop
pointsUnique = pointsUnique[~pointsUnique.intersects(unary_union(rooftopMerged.geometry))]

# Get points that were not georeferenced in GM-SEUS digitization effort
pointsUnique = pointsUnique[~pointsUnique.intersects(unary_union(gmseusDigGeoref_connectors.geometry))]
pointsUnique = pointsUnique.reset_index(drop=True)

# Print the resulting number of arrays and number of arrays needing manual digitization
print("Total unique point arrays in merged dataset: ", len(mergedPoints), " arrays")
print("Unique point arrays needing manual digitization: ", len(pointsUnique), " arrays")

# Export
pointsUnique.to_file(os.path.join(derivedTemp_path, r'points_toDigitize.shp'))
mergedPoints.to_file(os.path.join(derivedTemp_path, r'points_all.shp'))

# ~~~~~~~~~~~~~~~~~~~~~~~~~ Polygon Array Data

# Assign a global unique array ID to each array for processing
mergedMetadata['tmpArrID'] = range(1, len(mergedMetadata) + 1)

# Print the number of unique arrays
print(f'The number of unique arrays in the merged polygon dataset is {len(mergedMetadata)}')

# Print the total area in km2 of the arrays
print(f'The total area of arrays in the merged polygon dataset is {(mergedMetadata["area"].sum() / 1e6).round(2)} km2')

# Print value counts of Source column
print('Value counts of Source column in merged polygon dataset:')
print(mergedMetadata['Source'].value_counts())

# Export shapefile
mergedMetadata.to_file(os.path.join(derivedTemp_path, r'existingDatasetArrayShapes.shp'))
```

## Filter Panel-Rows by Priority and Ground-Mounted

```python
# Call existing rooftop array dataset
rooftopArraysInit = gpd.read_file(os.path.join(derivedTemp_path, r'existingRooftopArrayShapes.shp'))

# Create arrays from panels for all three datasets
gmseusDigPanels_arrays = gu.createArrayFromPanels(gmseusDigPanels, panelArrayBuff, '', '', False)
osmPanels_arrays = gu.createArrayFromPanels(osmPanels, panelArrayBuff, '', '', False)
ccvpvPanels_arrays = gu.createArrayFromPanels(ccvpvPanels, panelArrayBuff, '', '', False)

# Set arraySource column
gmseusDigPanels_arrays['arraySource'] = 'GMSEUSdigArraysPanels_v2_0'
osmPanels_arrays['arraySource'] = 'OSM'
ccvpvPanels_arrays['arraySource'] = 'CCVPV'

# Perform preferential spatial filtering on the arrays created from panels
panelsArrayPriorityList = [gmseusDigPanels_arrays, osmPanels_arrays, ccvpvPanels_arrays]
panelsArrayPriority = gu.preferentialSpatialFilter(panelsArrayPriorityList)

# Perform spatial join to get arraySource column attached to each panel-row dataset. Then, if panels source is equal to arraySource, keep the row. This will ensure that only panels that belong to the preferentially selected arrays are kept.
gmseusDigPanels_arraysJoined = gpd.sjoin(gmseusDigPanels, panelsArrayPriority[['arraySource', 'geometry']], how='left', predicate='intersects')
osmPanels_arraysJoined = gpd.sjoin(osmPanels, panelsArrayPriority[['arraySource', 'geometry']], how='left', predicate='intersects')
ccvpvPanels_arraysJoined = gpd.sjoin(ccvpvPanels, panelsArrayPriority[['arraySource', 'geometry']], how='left', predicate='intersects')
gmseusDigPanels_filtered = gmseusDigPanels_arraysJoined[gmseusDigPanels_arraysJoined['Source'] == gmseusDigPanels_arraysJoined['arraySource']]
osmPanels_filtered = osmPanels_arraysJoined[osmPanels_arraysJoined['Source'] == osmPanels_arraysJoined['arraySource']]
ccvpvPanels_filtered = ccvpvPanels_arraysJoined[ccvpvPanels_arraysJoined['Source'] == ccvpvPanels_arraysJoined['arraySource']]

# Combine the three panel datasets into a single geodataframe
mergedPanels = pd.concat([gmseusDigPanels_filtered, osmPanels_filtered, ccvpvPanels_filtered], ignore_index=True)
mergedPanels = gpd.GeoDataFrame(mergedPanels, geometry='geometry', crs=toCRS)

# Drop panels that overlap with rooftop arrays
mergedPanels = mergedPanels[~mergedPanels.intersects(unary_union(rooftopArraysInit.geometry))]

# Reset index
mergedPanels = mergedPanels.reset_index(drop=True)

# Drop temporary panel ID and arraySource columns, set new unique panel IDs
mergedPanels = mergedPanels.drop(columns=['tmpPnlID', 'arraySource', 'index_right'])
mergedPanels['tmpPnlID'] = range(1, len(mergedPanels) + 1)

# Drop panels with less area than the minimum panel area
mergedPanels = mergedPanels[mergedPanels['area'] >= minPanelRowArea]

# Print number of rows in mergedPanels
print(f'Number of final panel-rows in existing datasets: {len(mergedPanels)}')

# Print the total sum of 'area' in the mergedPanels dataset in km2
print(f'Total area of panels in mergedPanels dataset is {(mergedPanels["area"].sum() / 1e6).round(2)} km2')

# Print value counts of Source column
print('Value counts of Source column in merged panel dataset:')
print(mergedPanels['Source'].value_counts())

# Print number of rows in panelsArrayPriority
print(f'Number of unique arrays created from panels in existing datasets: {len(panelsArrayPriority)}')

# Print the total sum of 'area' in the panelsArrayPriority dataset in km2
print(f'Total area of arrays created from panels in existing datasets is {(panelsArrayPriority.geometry.area.sum() / 1e6).round(2)} km2')

# Print value counts of arraySource column
print('Value counts of arraySource column in panel-created array dataset:')
print(panelsArrayPriority['arraySource'].value_counts())

# Export
mergedPanels.to_file(os.path.join(derivedTemp_path, r'existingDatasetPanelShapes.shp'))
```
